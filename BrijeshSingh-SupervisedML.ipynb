##### a)	Describe the working of a Decision Tree algorithm. How does it decide on the best features to split the data? What are some advantages and disadvantages of using Decision Trees?

###### A Decision Tree is a supervised machine learning algorithm used for classification and regression. It works by recursively splitting the dataset into smaller subsets based on feature values, forming a tree‑like structure.

##### Working Steps
###### Start at the root node with the full dataset.

###### For every feature, the algorithm evaluates all possible split points.

###### It selects the best feature + best threshold that produces the purest child nodes.

###### The dataset is split into two or more branches.

###### The process repeats recursively on each child node (this is called recursive partitioning).

###### Splitting stops when:

###### - All samples in a node belong to one class

###### - Maximum depth is reached

###### - Minimum samples per node is reached

###### - No further improvement is possible

###### The final nodes are called leaf nodes, and each leaf represents a class label (classification) or a value (regression).

#### 2. How a Decision Tree Chooses the Best Feature to Split
###### A Decision Tree chooses the best feature using impurity reduction.
###### The goal is to create child nodes that are as pure as possible.

###### Common impurity measures
###### (a) Gini Impurity (CART – Classification and Regression Trees)

###### 𝐺𝑖𝑛𝑖
###### Lower Gini = purer node.

###### (b) Entropy (Information Gain – ID3, C4.5)
######    Higher Information Gain = better split.

###### (c) For Regression Trees
######    MSE (Mean Squared Error) reduction

######    MAE (Mean Absolute Error) reduction

###### 3. Advantages of Decision Trees
    - Easy to understand and interpret
    - Looks like human decision-making
    - Can be visualized graphically
    - No need for feature scaling
    - Works with raw data
    - No normalization/standardization required

    ✔ Handles both numerical and categorical data
    ✔ Captures non-linear relationships
    ✔ Requires little data preprocessing
    No need for dummy variables in some implementations

    Handles missing values (CART)

4. Disadvantages of Decision Trees
    ✘ Highly prone to overfitting
    Trees grow deep and memorize training data

    Needs pruning or max-depth control

    ✘ Unstable
        Small changes in data → completely different tree

    ✘ Biased toward features with many categories
        Categorical features with many levels may dominate splits

    ✘ Not great for continuous smooth predictions
        Regression trees produce step‑wise predictions
